```python
# First method to convert from object to float
df['income'].str.replace('$', '').str.replace(',', '').astype('float')
```

```python
def convert_to_numeric(x):

    if type(x) == float:
        return x
    
    else:
        return float(x.strip('$').replace(',', ''))
```

```python
df['income'].apply(convert_to_numeric)
```

```python
convert_cols = ['income', 'home_val', 'bluebook', 'oldclaim', 'clm_amt']

for col in convert_cols:
    df[col] = df[col].apply(convert_to_numeric)
```

# Feature Engineering

```python
# Extract new column for customer_loyalty from tif
df.tif
```

```python
df.tif.unique()
```

```python
def extract_customer_loyalty(x):

    if x <= 2:
        return 'New Customer'
    
    elif x <= 7:
        return 'Medium Loyalty'

    elif x <= 15:
        return 'Stable Loyalty'
    
    elif x <= 25:
        return 'High Loyalty'
    
df['customer_loyalty'] = df['tif'].apply(extract_customer_loyalty)
df['customer_loyalty']
```

```python
px.histogram(data_frame= df, x= 'customer_loyalty').update_xaxes(categoryorder = 'max descending')
```
